```python
import os

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import mlflow

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    root_mean_squared_error,
    mean_absolute_error,
    r2_score
)

print("All libraries imported successfully!")
```

```python
df = pd.read_csv(r"C:\Users\Umesh Chandra\Desktop\MLops-Day1\data\data.csv")
df.head()
```

```python
X =df[["TV","Radio","Newspaper"]]
y=df["Sales"]

Xtrain, Xtest, ytrain, ytest = train_test_split(X,y,test_size=0.2,random_state=56)
```

```python
mlflow.set_tracking_uri("sqlite:///mlflow.db")
```

```python
mlflow.set_experiment("Advertising Sales Prediction")
```

```python
with mlflow.start_run(run_name="LinearRegression"):

    model = LinearRegression()

    model.fit(Xtrain, ytrain)

    ypred = model.predict(Xtest)

    rmse = root_mean_squared_error(ytest, ypred)
    r2 = r2_score(ytest, ypred)

    # parameters
    mlflow.log_param("model_type","LinearRegression")
    mlflow.log_param("test_size",0.2)
    mlflow.log_param("random_state",56)

    #metrics
    mlflow.log_metric("test_rmse",rmse)
    mlflow.log_metric("test_r2",r2)
```

```python
with mlflow.start_run(run_name="RidgeRegression"):

    model = Ridge(alpha=1.0)

    model.fit(Xtrain, ytrain)

    ypred = model.predict(Xtest)

    rmse = root_mean_squared_error(ytest, ypred)
    r2 = r2_score(ytest, ypred)

    # parameters
    mlflow.log_param("model_type","RidgeRegression")
    mlflow.log_param("alpha",1.0)
    mlflow.log_param("test_size",0.2)
    mlflow.log_param("random_state",56)

    #metrics
    mlflow.log_metric("test_rmse",rmse)
    mlflow.log_metric("test_r2",r2)

    # Log Artifacts
    mlflow.sklearn.log_model(sk_model= model, name ="Ridge_Reg_model")
```
